# 03. Model Comparison Notebook

This notebook loads evaluation metrics generated across trained models (Baseline, Random Forest, XGBoost) and visualizes performance.

```python
import json
import pandas as pd
from ml.config.feature_config import METRICS_DIR

comp_file = METRICS_DIR / 'model_comparison.json'
if comp_file.exists():
    with open(comp_file, 'r') as f:
        data = json.load(f)
    df_metrics = pd.DataFrame(data['comparison'])
    print(f"Best Model: {data['best_model']}")
    display(df_metrics)
else:
    print("No model comparison metrics found yet. Run `python ml/training/train_models.py` first.")
```
