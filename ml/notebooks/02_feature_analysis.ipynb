# 02. Feature Analysis Notebook

This notebook analyzes feature correlations, numerical distributions, and engineered feature impacts.

```python
import pandas as pd
import numpy as np
from ml.config.feature_config import CLEANED_DATA_PATH, NUMERIC_COLS
from ml.preprocessing import FeatureEngineer

df = pd.read_csv(CLEANED_DATA_PATH)
fe = FeatureEngineer()
df_engineered = fe.transform(df)
print("Engineered columns added:", [c for c in df_engineered.columns if c not in df.columns])
```

```python
# Feature correlation with target
num_df = df_engineered[NUMERIC_COLS + ['total_engagement_score', 'pricing_intent_ratio', 'target']].dropna()
corrs = num_df.corr()['target'].sort_values(ascending=False)
print("Feature correlations with Target:")
print(corrs)
```
